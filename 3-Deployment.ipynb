```python
# Import required libraies
import pandas as pd
import pickle
```

```python
# Prepare user data
input_data = {
    "town": "BUKIT BATOK",
    "flat_type": "4 ROOM",
    "storey_range": "10 TO 12",
    "floor_area_sqm": 100,
    "flat_model": "Model A",
    "lease_commence_date": 2000,
    "house_age": 26,
    "price_per_sqm": 5500,
    "price_per_sqft": 500
}

input_data_df = pd.DataFrame([input_data])
```

```python
input_data_df
```

```python
training_columns = pickle.load(open("hdb_training_columns.sav", "rb"))
input_data_df = input_data_df.reindex(columns=training_columns, fill_value=0)
```

```python
input_data_df
```

```python
scaler_x = pickle.load(open("hdb_scaler_x.sav", "rb"))
scaler_y = pickle.load(open("hdb_scaler_y.sav", "rb"))
```

```python
scaler_x
```

```python
scaler_y
```

```python
scaled_data_x = scaler_x.transform(input_data_df)
```

```python
scaled_data_x
```

```python
# Load the finalized model
model = pickle.load(open("hdb_model_rf.sav", "rb"))
```

```python
model
```

```python
price = model.predict(scaled_data_x)
price
```

```python
price.shape
```

```python
inversed_price = scaler_y.inverse_transform(price.reshape(-1, 1))
inversed_price
```

```python
inversed_price.shape
```

```python
inversed_price.item()
```

```python
print(
    input_data["town"],  
    input_data["flat_type"], 
    input_data["storey_range"], 
    input_data["flat_model"], 
    input_data["lease_commence_date"], 
    "predicted price is", inversed_price.item())
```
