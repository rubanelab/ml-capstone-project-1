```python
from hopelib import HDB
hdb = HDB(dataset)
```

```python
# Get all towns in the dataset
hdb.get_all_towns()
```

```python
# Get all flat types
hdb.get_flat_types()
```

```python
# Get flat models
hdb.get_flat_models()
```

```python
# Grab the dataset for particular town only
bukit_batok_dataset = hdb.get_dataset_by_town("BUKIT BATOK")
```

```python
bukit_batok_dataset.head()
```

```python
# Average price in the town
price = hdb.get_average_price_by_town('BUKIT BATOK')
print("Average price in BUKIT BATOK: S$", round(price, 2))
```

```python
price = hdb.get_average_price_by_town('QUEENSTOWN')
print("Average price in QUEENSTOWN: S$", round(price, 2))
```

```python
# Highest price in the town
highest_price = hdb.get_highest_price_by_town('BUKIT BATOK')
print("Highest price in BUKIT BATOK: S$", round(highest_price, 2))
```

```python
# Lowest price in the town
highest_price = hdb.get_lowest_price_by_town('BUKIT BATOK')
print("Lowest price in BUKIT BATOK: S$", round(highest_price, 2))
```

```python
# Get available story range
storey_ranges = hdb.get_storey_ranges()
storey_ranges
```
